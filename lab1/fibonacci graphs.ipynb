```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set_theme(style="whitegrid")

fib_data = {
    "Algorithm": [
        "Recursive", "Recursive", "Recursive", "Recursive",
        "Iterative", "Iterative", "Iterative", "Iterative",
        "Dynamic Programming", "Dynamic Programming", "Dynamic Programming", "Dynamic Programming"
    ],
    "Value of n": [10, 20, 30, 40, 10, 20, 30, 40, 10, 20, 30, 40],
    "Execution Time (s)": [
        0.000028, 0.002427, 0.306202, 37.501368,
        0.000010, 0.000006, 0.000005, 0.000006,
        0.000026, 0.000015, 0.000013, 0.000013
    ],
    "Memory (Bytes)": [0, 0, 3350, 2869, 0, 0, 0, 0, 96, 176, 256, 336]
}

df_fib = pd.DataFrame(fib_data)

fig, ax1 = plt.subplots(figsize=(10, 6))
sns.lineplot(data=df_fib, x="Value of n", y="Execution Time (s)", hue="Algorithm", marker="o", linewidth=2.5, ax=ax1)
ax1.set_title("Fibonacci Algorithms: Value of n vs. Execution Time (Linear Scale)", fontweight="bold", pad=15)
ax1.set_xlabel("Value of n")
ax1.set_ylabel("Execution Time (Seconds)")
ax1.legend(title="Variants")
plt.tight_layout()
plt.show()

fig, ax2 = plt.subplots(figsize=(10, 6))
sns.lineplot(data=df_fib, x="Value of n", y="Execution Time (s)", hue="Algorithm", marker="o", linewidth=2.5, ax=ax2)
ax2.set_yscale('log')
ax2.set_title("Fibonacci Algorithms: Value of n vs. Execution Time (Log Scale)", fontweight="bold", pad=15)
ax2.set_xlabel("Value of n")
ax2.set_ylabel("Execution Time (Seconds, Log Scale)")
ax2.legend(title="Variants")
plt.tight_layout()
plt.show()

fig, ax3 = plt.subplots(figsize=(10, 6))
sns.lineplot(data=df_fib, x="Value of n", y="Memory (Bytes)", hue="Algorithm", marker="s", linewidth=2.5, ax=ax3)
ax3.set_title("Fibonacci Algorithms: Value of n vs. Memory Usage", fontweight="bold", pad=15)
ax3.set_xlabel("Value of n")
ax3.set_ylabel("Memory Consumption (Bytes)")
ax3.legend(title="Variants")
plt.tight_layout()
plt.show()
```
