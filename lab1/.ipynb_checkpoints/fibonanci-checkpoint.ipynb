```python
import time
import tracemalloc
import pandas as pd

inputs = [10, 20, 30, 40]
all_metrics = []
```

```python
def fibonacci_recursive(n):
    if n <= 1:
        return n
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

def get_recursive_ops(n):
    ops_memo = {0: 1, 1: 1}
    for i in range(2, n + 1):
        ops_memo[i] = ops_memo[i-1] + ops_memo[i-2] + 1
    return ops_memo[n]

print("--- RECURSIVE OUTPUTS ---")
for n in inputs:
    tracemalloc.start()
    start_time = time.perf_counter()
    
    result = fibonacci_recursive(n)
        
    end_time = time.perf_counter()
    _, peak_memory = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    
    print(f"Fibonacci number for n={n}: {result}")
    
    ops = get_recursive_ops(n)
    all_metrics.append({
        "Algorithm": "1. Recursive",
        "n": n,
        "Execution Time (s)": round(end_time - start_time, 6),
        "Memory (Bytes)": peak_memory,
        "Operations": ops
    })
```

```python
def fibonacci_iterative(n):
    first = 0
    second = 1
    for _ in range(n):
        first, second = second, first + second
    return first

print("--- ITERATIVE OUTPUTS ---")
for n in inputs:
    tracemalloc.start()
    start_time = time.perf_counter()
    
    result = fibonacci_iterative(n)
        
    end_time = time.perf_counter()
    _, peak_memory = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    
    print(f"Fibonacci number for n={n}: {result}")
    
    all_metrics.append({
        "Algorithm": "2. Iterative",
        "n": n,
        "Execution Time (s)": round(end_time - start_time, 6),
        "Memory (Bytes)": peak_memory,
        "Operations": n
    })
```

```python
def fibonacci_dynamic(n):
    table = [0] * (n + 1)
    if n >= 1:
        table[1] = 1
    for i in range(2, n + 1):
        table[i] = table[i - 1] + table[i - 2]
    return table[n]

print("--- DYNAMIC PROGRAMMING OUTPUTS ---")
for n in inputs:
    tracemalloc.start()
    start_time = time.perf_counter()
    
    result = fibonacci_dynamic(n)
        
    end_time = time.perf_counter()
    _, peak_memory = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    
    print(f"Fibonacci number for n={n}: {result}")
    
    all_metrics.append({
        "Algorithm": "3. Dynamic Programming",
        "n": n,
        "Execution Time (s)": round(end_time - start_time, 6),
        "Memory (Bytes)": peak_memory,
        "Operations": n - 1 if n >= 2 else 0
    })
```

```python
df_comparison = pd.DataFrame(all_metrics)
df_comparison = df_comparison.sort_values(by=["n", "Algorithm"]).reset_index(drop=True)
df_comparison.to_csv("fibonacci_final_comparison.csv", index=False)

print("--- FINAL METRICS COMPARISON ---")
df_comparison
```
