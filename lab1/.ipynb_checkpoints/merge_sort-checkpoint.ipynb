```python
import time
import tracemalloc
import random
import sys

sys.setrecursionlimit(20000)

def merge_sort(values):
    if len(values) <= 1:
        return values

    middle = len(values) // 2
    left = merge_sort(values[:middle])
    right = merge_sort(values[middle:])

    result = []
    left_index = 0
    right_index = 0

    while left_index < len(left) and right_index < len(right):
        if left[left_index] <= right[right_index]:
            result.append(left[left_index])
            left_index += 1
        else:
            result.append(right[right_index])
            right_index += 1

    result.extend(left[left_index:])
    result.extend(right[right_index:])

    return result

def run_automated_test(size, input_type):
    tracemalloc.start()
    
    if input_type == "Sorted":
        test_data = list(range(size))
    elif input_type == "Reverse-Sorted":
        test_data = list(range(size, 0, -1))
    elif input_type == "Random":
        test_data = random.sample(range(size * 10), size)
    
    start_time = time.time()
    merge_sort(test_data) 
    
    end_time = time.time()
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    
    execution_time = end_time - start_time
    peak_memory_mb = peak / (1024 * 1024) 
    
    print(f"{input_type:<15} | Size: {size:<6} | Time: {execution_time:.6f}s | Peak Memory: {peak_memory_mb:.6f} MB")

print("Choose your execution mode:")
print("1: Manual Input (Type your own numbers)")
print("2: Automated Benchmark (Test all sizes & conditions)")
choice = input("Enter 1 or 2: ").strip()

if choice == "1":
    user_input = input("\nEnter numbers separated by spaces: ")
    data = list(map(int, user_input.split()))
    
    tracemalloc.start()
    start_time = time.time()
    
    sorted_data = merge_sort(data)
    
    end_time = time.time()
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    
    print("\n--- Manual Run Results ---")
    print("Sorted list:", sorted_data)
    print(f"Time Taken  : {(end_time - start_time):.6f}s")
    print(f"Peak Memory : {(peak / (1024 * 1024)):.6f} MB")

elif choice == "2":
    input_sizes = [100, 500, 1000, 5000, 10000]
    input_types = ["Sorted", "Reverse-Sorted", "Random"]

    print("\nExecuting Merge Sort benchmarks...\n")
    for size in input_sizes:
        for itype in input_types:
            run_automated_test(size, itype)
else:
    print("Invalid choice. Please re-run the cell and select 1 or 2.")
```
