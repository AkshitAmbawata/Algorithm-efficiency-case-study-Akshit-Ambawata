```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 11, 'axes.labelsize': 12, 'axes.titlesize': 14})

sorting_data = {
    "Algorithm": [
        "Bubble Sort", "Bubble Sort", "Bubble Sort", "Bubble Sort",
        "Insertion Sort", "Insertion Sort", "Insertion Sort", "Insertion Sort",
        "Merge Sort", "Merge Sort", "Merge Sort", "Merge Sort", "Merge Sort",
        "Quick Sort", "Quick Sort", "Quick Sort", "Quick Sort", "Quick Sort"
    ],
    "Input Size": [
        100, 500, 1000, 5000,
        100, 500, 1000, 5000,
        100, 500, 1000, 5000, 10000,
        100, 500, 1000, 5000, 10000
    ],
    "Execution Time (s)": [
        0.000715, 0.067400, 0.438733, 12.417061,
        0.000338, 0.030265, 0.210696, 5.773154,
        0.000909, 0.004579, 0.010646, 0.084550, 0.205042,
        0.000676, 0.004245, 0.007150, 0.038132, 0.084170
    ],
    "Peak Memory (MB)": [
        0.028526, 0.061089, 0.095451, 0.959394,
        0.028923, 0.061455, 0.095940, 0.182783,
        0.028923, 0.061546, 0.095634, 0.959007, 1.182365,
        0.028923, 0.061058, 0.095604, 0.958427, 1.182243
    ]
}

df_sort = pd.DataFrame(sorting_data)

fig, ax1 = plt.subplots(figsize=(10, 6))
sns.lineplot(data=df_sort, x="Input Size", y="Execution Time (s)", hue="Algorithm", marker="o", linewidth=2.5, ax=ax1)
ax1.set_title("Sorting Algorithms: Input Size vs. Execution Time", fontweight="bold", pad=15)
ax1.set_xlabel("Input Size (N)")
ax1.set_ylabel("Execution Time (Seconds)")
ax1.legend(title="Algorithms")
plt.tight_layout()
plt.show()

fig, ax2 = plt.subplots(figsize=(10, 6))
sns.lineplot(data=df_sort, x="Input Size", y="Peak Memory (MB)", hue="Algorithm", marker="s", linewidth=2.5, ax=ax2)
ax2.set_title("Sorting Algorithms: Input Size vs. Peak Memory Usage", fontweight="bold", pad=15)
ax2.set_xlabel("Input Size (N)")
ax2.set_ylabel("Peak Memory Consumption (MB)")
ax2.legend(title="Algorithms")
plt.tight_layout()
plt.show()
```
